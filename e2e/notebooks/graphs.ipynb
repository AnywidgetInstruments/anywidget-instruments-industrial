```python
import numpy as np
import anywidget_instruments as ai
chart = ai.WaveformChart(label='E2E graph', history=100, dt=1, y_min=0, y_max=100, size=(420, 220))
chart.append(np.arange(100.0))
echo = ai.SevenSegment(0, digits=4, decimals=0, label='E2E cursor value')
chart.observe(lambda c: setattr(echo, 'value', c['new'][0]['values'][0] if c['new'] else -1), names='cursor_values')
chart.add_cursor(20)
ai.Panel([chart, echo], columns=2)
```
