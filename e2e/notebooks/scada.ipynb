```python
import anywidget_instruments_industrial as ai
pump = ai.Pump(tag='P-1', label='E2E pump', simulate=True, auto=False)
led = ai.LED(label='E2E running')
pump.on_command(lambda e: setattr(led, 'value', e['command'] == 'start'))
syn = ai.SynopticCanvas(label='E2E synoptic', size=(420, 260))
syn.add(ai.Tank(40, label='E2E nested tank'), x=10, y=10)
syn.add(pump, x=200, y=20)
syn.add(led, x=320, y=40)
syn.add_pipe([(130, 120), (230, 120)], flow=True)
syn
```

```python
banner = ai.AlarmBanner(label='E2E banner')
acked = ai.LED(label='E2E acked')
banner.on_acknowledge(lambda e: setattr(acked, 'value', True))
banner.raise_alarm('TK-1.HI', 'Level high', source='TK-1')
ai.Panel([banner, acked], columns=2)
```
