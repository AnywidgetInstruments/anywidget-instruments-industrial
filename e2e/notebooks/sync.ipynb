```python
import anywidget_instruments_industrial as ai

knob = ai.Knob(10, step=1, label="E2E knob")
knob
```

```python
led = ai.LED(label="E2E led")
echo = ai.SevenSegment(0, digits=4, decimals=0, label="E2E echo")
gauge = ai.Gauge(0, label="E2E gauge")
ai.Panel([led, echo, gauge])
```

```python
# front end -> kernel: the knob change is echoed back by the kernel
@knob.on_change
def _echo(change):
    led.value = True
    echo.value = change["new"]
```

```python
# kernel -> front end after display
gauge.value = 77
```

```python
import numpy as np

chart = ai.WaveformChart(label="E2E chart", history=200)
display(chart)
chart.append(np.linspace(-1, 0.5, 150))
```

```python
pic = ai.PictureControl(label="E2E picture", size=(200, 100))
clicks = ai.SevenSegment(0, digits=4, decimals=0, label="E2E clicks")
pic.on_click(lambda e: setattr(clicks, "value", round(e["x"])))
pic.rect(0, 0, 50, 50, stroke=None, fill="#ff0000")
pic.text(60, 30, "hello")
ai.Panel([pic, clicks])
```
